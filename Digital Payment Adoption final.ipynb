## A) HANDLING THE MISSING VALUES

This dataset has 39 null values which is numerical data. It's present in the columns like atm onsite/offsite, credit card based columns.

```python


#identify columns with missing values that need imputation
columns_to_impute = ['atms_crms_onsite','atms_crms_offsite','cc_pay_trns_at_pos_vol','cc_pay_trns_at_pos_val','cc_pay_trns_online_vol','cc_pay_trns_online_val','cc_pay_trns_other_val','cc_pay_trns_other_vol','cc_cash_withdraw_atm_val','cc_cash_withdraw_atm_vol']

for col in columns_to_impute:
    #calculate the median for the current column
    median_value = df[col].median()

    #fill null values in the current column with its calculated median
    df[col]= df[col].fillna(median_value)
    print(f"Filled missing values in '{col}' with median:{median_value:.2f}")
```

```python
#Total no.of.missing vales after imputation.
df.isnull().sum()
```

NB : The median is chosen because it is less sensitive to outliers compared to the mean, making it a robust choice for imputing numerical data.

## C) FEATURE ENGINEERING
In this phase the project required further addition of data range/ new columns for a profound analysis.

```python
#Existing Columns
print('Existing Columns:', df.columns.tolist())
```

```python
# Create Month and Year

df['month'] = df['date'].dt.month
df['year'] = df['date'].dt.year
df.head(2)
```

```python
#total atm count.

df['total_atm_count'] = (
    df['atms_crms_onsite'] +
    df['atms_crms_offsite']
)
df.head(2)
```

```python
# atm infrastructure level

df['infrastructure_level'] = pd.qcut(
    df['total_atm_count'],
    q=3,
    labels=['Low', 'Moderate', 'High']
)
df.head(2)
```

```python
# Total credit card transaction volume to show no.of transaction done

df['total_credit_card_transaction_volume'] = (
    df['cc_pay_trns_at_pos_vol'] +
    df['cc_pay_trns_online_vol'] +
    df['cc_pay_trns_other_vol']
)

# Total debit card transaction volume to show no.of transaction done

df['total_debit_card_transaction_volume'] = (
    df['dc_pay_trns_at_pos_vol'] +
    df['dc_pay_trns_online_vol'] +
    df['dc_pay_trns_other_vol']
)
df.head(2)
```

```python
# Total credit card transaction value to show no.of. transaction amount.

df['total_credit_card_transaction_value'] = (
    df['cc_pay_trns_at_pos_val'] +
    df['cc_pay_trns_online_val'] +
    df['cc_pay_trns_other_val']
)

# Total debit card transaction value to show no.of. transaction amount.

df['total_debit_card_transaction_value'] = (
    df['dc_pay_trns_at_pos_val'] +
    df['dc_pay_trns_online_val'] +
    df['dc_pay_trns_other_val']
)
df.head(2)
```

```python
#Credit card usage type cash/ digital

#Digital credit-card volume:
credit_digital = (
    df['cc_pay_trns_at_pos_vol'] +
    df['cc_pay_trns_online_vol'] +
    df['cc_pay_trns_other_vol']
)

#Cash credit-card volume:
credit_cash = df['cc_cash_withdraw_atm_vol']

df['credit_card_usage_type'] = np.where(
    credit_digital >= credit_cash,
    'Digital',
    'Cash'
)

#Debit-card digital transaction volume:
debit_digital = (
    df['dc_pay_trns_at_pos_vol'] +
    df['dc_pay_trns_online_vol'] +
    df['dc_pay_trns_other_vol']
)

#Debit-card cash withdrawal volume:
debit_cash = (
    df['dc_cash_withdraw_atm_vol'] +
    df['dc_cash_withdraw_pos_vol']
)

df['debit_card_usage_type'] = np.where(
    debit_digital >= debit_cash,
    'Digital',
    'Cash'
)

df.head(2)
```

```python
# Average  of credit card transaction value which calculate avg spending per cc.

df['average_credit_card_transaction_value'] = (
    df['total_credit_card_transaction_value'] /
    df['total_credit_card_transaction_volume']
)

# Average of debit card transaction value which calculate avg spending per dc.
df['average_debit_card_transaction_value']= (
    df['total_debit_card_transaction_value'] /
    df['total_debit_card_transaction_volume']
)
df.head(2)
```

```python
#Filling the Nan values to 0

df['average_credit_card_transaction_value'] = df['average_credit_card_transaction_value'].fillna(0)
df.head(2)
```

Average Credit/Debit card transaction value that calculates average spending per CC/DC. So by calculating the average I got some 'Nan' values as a result. This 'Nan' values appears in the average transaction column due to zero transaction volumes for certain banks, and this was handled by replacing '0'. 

```python
#Total QR infrastructure

df['total_qr_infrastructure'] = (
    df['bharat_qr'] +
    df['upi_qr']
)
df.head(2)
```

```python
# Digital indrastructure index to show how digitaly equipped the banks are.

digital_infrastructure_index = [
    'pos',
    'micro_atms',
    'bharat_qr',
    'upi_qr',
    'total_atm_count'
]

df['digital_infrastructure_index'] = (
    df[digital_infrastructure_index] /
    df[digital_infrastructure_index].max()
).mean(axis=1)
df.head(2)
```

```python
# Digital infrastructure level to show the infrastructure level of banks.

df['digital_infrastructure_level'] = pd.qcut(
    df['digital_infrastructure_index'],
    q=3,
    labels=['Low', 'Moderate', 'High']
)
df.head(2)
```

```python
# QR preferences by banks

df['QR_preference'] = np.where(
    df['upi_qr'] > df['bharat_qr'],
    'UPI QR',
    'Bharat QR'
)
df.head(2)
```

```python
# Mcro atm dependency level that shows banks with micro atm presence

df['micro_atm_dependency_level'] = np.where(
    df['micro_atms'] > 0,
    'Present',
    'Not Present'
)
df.head(2)
```

1. Shift Focus to UPI QR:

   Prioritize UPI QR expansion over legacy Bharat QR, as UPI dominates merchant adoption and volume.

2. Optimize Physical ATM Networks: 

   Rationalize underperforming offsite ATMs and shift capital toward high-yield, high-footfall locations.

3. Double Down on Micro ATMs for Rural Reach: 

   Scale Micro ATM distribution in Public Sector, Payment, and Small Finance banks to drive financial inclusion in semi-urban and rural regions.

4. Expand Credit Card Merchant Infrastructure:

   Invest in Point-of-Sale (POS) terminals specifically targeted at high-value Credit Card transactions, where growth remains strong.

5. Phase Out Debit Card POS Strategies:

   Reduce capital spending on physical Debit Card POS hardware, as consumers increasingly replace debit cards with UPI for retail payments.

6. Deploy AI-Driven Fraud Monitoring: 

   Implement real-time risk scoring and automated monitoring on high-volume QR channels to prevent cyber fraud and build transaction trust.

7. Cross-Sell Digital Services via Micro ATMs:

   Equip Micro ATM agents with capabilities to disburse micro-loans, insurance, and government subsidy cash-outs.

8. Bridge Digital Infrastructure Gaps in SFBs:

   Modernize Small Finance and Foreign Bank infrastructure from "Low" to "Moderate/High" tiers to stay competitive.

9. Promote Interoperable Merchant Terminals:

   Deploy hybrid devices that integrate POS, UPI QR display, and Aadhaar-enabled biometric cash withdrawal into a single terminal.

10. Target Private Bank Merchant Partnerships: 

    Public sector banks should partner with private acquirers to boost their low POS presence in urban retail hubs.
